```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

## Barplot

The **barplot** is one of the most commonly used categorical plots in Seaborn. It provides a straightforward way to visualize the average (or other estimator) of a quantitative variable for different categories. Essentially, a barplot shows the central tendency of the data for each category along with confidence intervals, making it easy to compare groups at a glance.

### Key Features:
- **Displays central tendency:** By default, barplot shows the mean value of the numerical variable for each category.
- **Customizable estimator:** You can change the aggregation function (e.g., median, sum) with the `estimator` parameter.
- **Confidence intervals:** The plot includes error bars to represent uncertainty, which can be customized or removed.
- **Hue parameter:** Allows grouping by a secondary categorical variable, adding color-coded bars for subgroups.
- **Orientation:** Bars can be vertical or horizontal via the `orient` parameter.
- **Data flexibility:** Works well with DataFrames and allows for direct passing of x, y, and hue columns.


### Tips for Using Barplots:

* Use barplots when you want to summarize data and compare averages or sums across categories.
* Be cautious with barplots on small datasets, as the average and confidence intervals might be misleading.
* When comparing distributions or spread, consider combining barplots with other plots like `stripplot` or `swarmplot` for added detail.

```python
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Average Total Bill per Day')
plt.show()
```

```python
sns.barplot(x='day', y='total_bill', data=tips, estimator=sum)
plt.title('Total Bill Sum per Day')
plt.show()
```

```python
sns.barplot(x='day', y='total_bill', hue='sex', data=tips)
plt.title('Average Total Bill per Day by Sex')
plt.show()
```

```python
sns.barplot(x='day', y='total_bill', data=tips, errorbar=None)
plt.title('Average Total Bill per Day (No Confidence Interval)')
plt.show()
```

## Countplot

The **countplot** is a powerful and intuitive plot for visualizing the frequency of observations in each category of a categorical variable. Unlike the barplot, which shows the average of a numerical variable, countplot directly displays the count of records for each category — making it perfect for understanding the distribution of categorical data.

### Key Features:
- **Displays counts:** Automatically counts the number of occurrences for each category in the dataset.
- **Simple and fast:** Great for quick exploration of categorical distributions.
- **Hue support:** Allows splitting the counts further by a secondary categorical variable, making subgroup comparisons easy.
- **Orientation:** Can be plotted vertically or horizontally.
- **Flexible input:** Accepts a categorical variable as `x` or `y` and data in DataFrame format.

### Tips for Using Countplots:

* Use countplots when you want to quickly see the frequency distribution of a categorical variable.
* When combined with `hue`, countplots can reveal how different subgroups contribute to the overall distribution.
* If you have many categories or sparse data, consider rotating labels or using horizontal orientation for readability.

Countplots are an essential tool for exploratory data analysis and provide a clear picture of categorical data distribution with minimal setup.


```python
sns.countplot(x='day', data=tips)
plt.title('Number of Records per Day')
plt.show()
```

```python
sns.countplot(x='day', hue='sex', data=tips)
plt.title('Number of Records per Day by Sex')
plt.show()
```

```python
sns.countplot(y='day', data=tips)
plt.title('Number of Records per Day (Horizontal)')
plt.show()
```

## Boxplot

The **boxplot** (also known as a box-and-whisker plot) is a statistical visualization that displays the distribution, spread, and potential outliers of a numerical variable across different categories. It is especially useful for comparing distributions between groups.

In Seaborn, `boxplot()` makes it simple to generate boxplots and includes many customization options to reveal important aspects of your data.

### Boxplot Parameter

| Parameter       | Description                                                                 | Example Usage                          |
|----------------|-----------------------------------------------------------------------------|----------------------------------------|
| `x` / `y`       | Categorical and numerical variables for the axes                           | `x='day', y='total_bill'`              |
| `hue`           | Categorical variable to split the data within each category                | `hue='sex'`                            |
| `data`          | DataFrame containing the data                                               | `data=tips`                            |
| `orient`        | Orientation of the plot: `'v'` (vertical) or `'h'` (horizontal)            | `orient='h'`                           |
| `palette`       | Color palette for the boxes                                                 | `palette='Set2'`                       |
| `width`         | Width of the boxes (0 to 1)                                                 | `width=0.6`                            |
| `linewidth`     | Width of the box edge lines                                                 | `linewidth=2`                          |
| `fliersize`     | Size of the outlier markers                                                 | `fliersize=3`                          |
| `showfliers`    | Whether to display outlier points                                           | `showfliers=False`                     |
| `notch`         | Whether to draw a notch to show confidence interval around median          | `notch=True`                           |
| `order`         | Specific order of categories on the axis                                   | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`     | Order for the hue category levels                                           | `hue_order=['Female', 'Male']`         |
| `dodge`         | When using `hue`, whether to separate the boxes for each level             | `dodge=True` (default)                 |
| `saturation`    | Transparency of the boxes (0 to 1)                                          | `saturation=0.75`                      |
| `ax`            | Matplotlib axis object to draw the plot onto                               | `ax=ax`                                |



### Tips for Using Boxplots:

* Ideal for comparing **distributions** and **detecting outliers** across groups.
* Combine with swarmplots or stripplots to overlay individual data points.
* Use with caution when sample sizes are small — outlier detection may be misleading.
* Customize whisker lengths using the `whis` parameter (default is 1.5 \* IQR).

Boxplots provide a concise summary of the data’s distribution and are an excellent tool for spotting patterns, variability, and potential anomalies within categories.


```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Distribution of Total Bill by Day')
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', hue='sex', data=tips)
plt.title('Total Bill Distribution by Day and Sex')
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips,hue='sex', palette='Set2')
plt.title('Styled Boxplot of Total Bill by Day')
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, showfliers=False)
plt.title('Total Bill by Day (Without Outliers)')
plt.show()
```

## Violinplot

The **violinplot** combines aspects of both a **boxplot** and a **kernel density plot**, offering a deeper understanding of the distribution of a numerical variable across categories. While boxplots summarize distributions through quartiles, violinplots reveal the **full probability distribution**, making them especially useful when the data is multimodal or skewed.

In Seaborn, `violinplot()` makes it easy to visualize both central tendency and distribution shape across different groups.


### Violinplot Parameter Table

| Parameter    | Description                                                                           | Example Usage                         |
| ------------ | ------------------------------------------------------------------------------------- | ------------------------------------- |
| `x` / `y`    | Categorical and numerical variables for axes                                          | `x='day', y='total_bill'`             |
| `hue`        | Categorical variable for subgrouping                                                  | `hue='sex'`                           |
| `data`       | DataFrame containing the input data                                                   | `data=tips`                           |
| `order`      | Specific order for categorical axis levels                                            | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`  | Specific order for hue levels                                                         | `hue_order=['Female', 'Male']`        |
| `palette`    | Color palette for violins                                                             | `palette='pastel'`                    |
| `split`      | Draw multiple hue levels on the same side of each violin                              | `split=True`                          |
| `scale`      | Method for scaling violins: `'area'`, `'count'`, `'width'`                            | `scale='width'`                       |
| `scale_hue`  | Whether to scale the violins by the number of observations per hue                    | `scale_hue=False`                     |
| `inner`      | Representation inside violins: `'box'`, `'quartile'`, `'point'`, `'stick'`, or `None` | `inner='quartile'`                    |
| `bw`         | Bandwidth of the KDE                                                                  | `bw=0.2`                              |
| `cut`        | Extends the KDE past the extreme datapoints (default is 2)                            | `cut=0`                               |
| `linewidth`  | Width of the violin edges                                                             | `linewidth=2`                         |
| `width`      | Width of the violin                                                                   | `width=0.8`                           |
| `orient`     | Orientation: `'v'` (vertical) or `'h'` (horizontal)                                   | `orient='h'`                          |
| `saturation` | Transparency of the colors (0 to 1)                                                   | `saturation=0.75`                     |
| `ax`         | Matplotlib axis to draw on                                                            | `ax=ax`                               |

---

### Tips for Using Violinplots:

* Use `violinplot` when you want more **detailed insights** into data distribution than a boxplot can provide.
* Combine with `stripplot` or `swarmplot` to overlay individual data points.
* Consider turning off the KDE tail extension with `cut=0` if you want to keep violins tightly bound to your data.


```python
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Distribution of Total Bill by Day')
plt.show()
```

```python
sns.violinplot(x='day', y='total_bill', hue='sex', data=tips, split=True)
plt.title('Distribution of Total Bill by Day and Sex')
plt.show()
```

## Stripplot

The **stripplot** is a simple and effective way to visualize **all individual data points** in a dataset, grouped by categories. Unlike boxplots or violinplots that summarize distributions, stripplots display **every observation**, making them ideal for understanding the raw data and spotting outliers or clusters.

In Seaborn, `stripplot()` is often used alone or layered on top of other plots (like boxplots or violinplots) to provide additional context.


### Stripplot Parameter Reference Table

| Parameter   | Description                                                            | Example Usage                         |
| ----------- | ---------------------------------------------------------------------- | ------------------------------------- |
| `x` / `y`   | Categorical and numerical variables                                    | `x='day', y='total_bill'`             |
| `hue`       | Categorical variable for subgrouping                                   | `hue='sex'`                           |
| `data`      | DataFrame containing the data                                          | `data=tips`                           |
| `order`     | Order of the categorical variable levels                               | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order` | Order of hue levels                                                    | `hue_order=['Female', 'Male']`        |
| `jitter`    | Adds random noise to prevent overlap (`True`, `False`, or float value) | `jitter=0.2`                          |
| `dodge`     | Separates the points for each level of the hue variable                | `dodge=True`                          |
| `palette`   | Color palette for hue categories                                       | `palette='Set1'`                      |
| `size`      | Size of the dots                                                       | `size=5`                              |
| `linewidth` | Border width of the dots                                               | `linewidth=0.5`                       |
| `edgecolor` | Border color of the dots                                               | `edgecolor='gray'`                    |
| `alpha`     | Transparency level (0 to 1)                                            | `alpha=0.7`                           |
| `orient`    | Orientation: `'v'` (vertical) or `'h'` (horizontal)                    | `orient='h'`                          |
| `ax`        | Matplotlib axis to draw the plot on                                    | `ax=ax`                               |

---

### Tips for Using Stripplots:

* Use `stripplot` when you want to show **every data point** in the distribution.
* Combine with `boxplot` or `violinplot` for richer visualizations.
* Use `jitter` or `dodge` to enhance readability when data points overlap.
* Keep in mind that for large datasets, `stripplot` can become cluttered — use `swarmplot` instead for automatic point adjustment.

The stripplot is a minimal, yet insightful tool for visualizing individual observations in categorical datasets.

```python
sns.stripplot(x='day', y='total_bill', data=tips)
plt.title('Total Bill Distribution by Day (Stripplot)')
plt.show()
```

```python
sns.stripplot(x='day', y='total_bill', data=tips, jitter=True)
plt.title('Stripplot with Jitter')
plt.show()
```

```python
sns.stripplot(x='day', y='total_bill', hue='sex', data=tips, dodge=True)
plt.title('Stripplot by Day and Sex')
plt.show()
```

## Swarmplot

The **swarmplot** is similar to the **stripplot**, but with a major advantage: it automatically arranges data points to avoid overlap, making the visualization more readable. It’s perfect for showing **all individual data points** in a category while maintaining a clear view of **density and distribution**.

In Seaborn, `swarmplot()` is a go-to tool for **visualizing small-to-medium datasets** where you want to show every observation without clutter.

---

### Swarmplot Parameter Table

| Parameter     | Description                                         | Example Usage                         |
| ------------- | --------------------------------------------------- | ------------------------------------- |
| `x` / `y`     | Categorical and numerical variables                 | `x='day', y='total_bill'`             |
| `hue`         | Categorical variable for subgrouping                | `hue='sex'`                           |
| `data`        | DataFrame containing the input data                 | `data=tips`                           |
| `order`       | Order of categorical axis values                    | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`   | Specific order for hue levels                       | `hue_order=['Female', 'Male']`        |
| `dodge`       | Separate hue categories to avoid overlap            | `dodge=True`                          |
| `palette`     | Color palette for hue levels                        | `palette='husl'`                      |
| `size`        | Size of the markers                                 | `size=5`                              |
| `linewidth`   | Width of marker borders                             | `linewidth=0.5`                       |
| `edgecolor`   | Edge color of markers                               | `edgecolor='gray'`                    |
| `alpha`       | Transparency of markers (0 to 1)                    | `alpha=0.7`                           |
| `orient`      | Orientation of the plot: `'v'` or `'h'`             | `orient='h'`                          |
| `ax`          | Matplotlib axis object to draw the plot             | `ax=ax`                               |
| `warn_thresh` | Threshold for performance warning on large datasets | `warn_thresh=1000`                    |

---

### Tips for Using Swarmplots:

* Use `swarmplot` when you want to show **all individual data points clearly** without overlap.
* Ideal for **small to medium-sized datasets** — it may become slow or cluttered with large datasets.
* Combine with `boxplot` or `violinplot` to create **rich visualizations** that show both summaries and individual points.
* Avoid using `swarmplot` when you have more than a few thousand data points — it becomes computationally expensive.

Swarmplots give you the best of both worlds: clear visibility of every point, and clean separation across categories.

```python
sns.swarmplot(x='day', y='total_bill', data=tips)
plt.title('Total Bill Distribution by Day (Swarmplot)')
plt.show()
```

```python
sns.swarmplot(x='day', y='total_bill', hue='sex', data=tips, dodge=True)
plt.title('Swarmplot by Day and Sex')
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, whis=[0, 100], width=0.6, palette='pastel')
sns.swarmplot(x='day', y='total_bill', data=tips, color='0.25')
plt.title('Boxplot with Swarmplot Overlay')
plt.show()
```

## Catplot (Categorical Plot FacetGrid)

The **catplot** function is a powerful and flexible interface in Seaborn that unifies all categorical plots — such as **barplot**, **boxplot**, **violinplot**, **stripplot**, and **swarmplot** — under one command. It allows easy **faceting** (splitting into subplots), supports **subgrouping**, and provides a cleaner, more modular way to build complex visualizations.

Catplot replaces the older `factorplot` and is now the **recommended way** to create multi-plot grids for categorical data.

---

### Catplot Parameter Reference Table

| Parameter           | Description                                                                 | Example Usage                         |
| ------------------- | --------------------------------------------------------------------------- | ------------------------------------- |
| `x` / `y`           | Categorical and numerical variables                                         | `x='day', y='total_bill'`             |
| `hue`               | Subgrouping variable (colors categories)                                    | `hue='sex'`                           |
| `col` / `row`       | Variables to create facet grids (columns or rows)                           | `col='smoker'`, `row='time'`          |
| `data`              | DataFrame to use for plotting                                               | `data=tips`                           |
| `kind`              | Type of plot: `'strip'`, `'swarm'`, `'box'`, `'violin'`, `'bar'`, `'count'` | `kind='violin'`                       |
| `order`             | Custom order for category levels                                            | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`         | Custom order for hue categories                                             | `hue_order=['Female', 'Male']`        |
| `height`            | Height of each facet                                                        | `height=4`                            |
| `aspect`            | Aspect ratio of each facet                                                  | `aspect=1.2`                          |
| `margin_titles`     | Place row/col titles on the margins                                         | `margin_titles=True`                  |
| `palette`           | Color palette                                                               | `palette='husl'`                      |
| `sharex` / `sharey` | Share axis scale across facets                                              | `sharex=False`, `sharey=True`         |
| `legend`            | Show legend or not                                                          | `legend=False`                        |
| `legend_out`        | Draw the legend outside the plot                                            | `legend_out=False`                    |

---

### Tips for Using Catplot:

* Use `catplot` for **faceted, multi-plot layouts** when comparing across multiple categories.
* Choose the appropriate `kind` depending on your visualization needs.
* Tweak `height`, `aspect`, and `col_wrap` for responsive, clean grid layouts.
* Use `catplot().set_titles()`, `set_axis_labels()`, etc., for further customization via the returned `FacetGrid`.

```python
sns.catplot(x='day', y='total_bill', kind='box', data=tips)
plt.suptitle('Boxplot of Total Bill by Day', y=1.02)
plt.show()
```

```python
sns.catplot(x='day', y='total_bill', hue='sex', col='smoker',
            kind='swarm', data=tips, height=4, aspect=0.9)
plt.suptitle('Total Bill by Day and Sex, Faceted by Smoker', y=1.05)
plt.show()
```

# Matrix Plots

Matrix plots allow you to plot data as color-encoded matrices and can also be used to indicate clusters within the data (later in the machine learning section we will learn how to formally cluster data).

```python
flights = sns.load_dataset('flights')
flights.head()
```

### Heatmap

A **heatmap** is a two-dimensional visualization where the individual values contained in a matrix are represented as colors. It's perfect for displaying relationships between two categorical variables or for visualizing correlation matrices.

In Seaborn, the `heatmap()` function allows you to create highly customizable and aesthetically pleasing heatmaps with features like annotations, colorbars, value formatting, and more.

---

#### Heatmap Parameter Reference Table

| Parameter                     | Description                                       | Example Usage                                 |
| ----------------------------- | ------------------------------------------------- | --------------------------------------------- |
| `data`                        | 2D array-like data (e.g., DataFrame, matrix)      | `data=corr`                                   |
| `annot`                       | If `True`, write the data values in each cell     | `annot=True`                                  |
| `fmt`                         | String format code for annotations                | `fmt=".2f"`                                   |
| `cmap`                        | Colormap for the heatmap                          | `cmap='coolwarm'`                             |
| `linewidths`                  | Width of the lines between cells                  | `linewidths=0.5`                              |
| `linecolor`                   | Color of the lines between cells                  | `linecolor='white'`                           |
| `vmin` / `vmax`               | Minimum and maximum values for colormap scaling   | `vmin=0, vmax=1`                              |
| `center`                      | The value at which to center the colormap         | `center=0`                                    |
| `square`                      | Force cells to be square-shaped                   | `square=True`                                 |
| `xticklabels` / `yticklabels` | Show/hide or customize tick labels                | `xticklabels=True`, `yticklabels=['Mon',...]` |
| `cbar`                        | Show/hide the color bar                           | `cbar=False`                                  |
| `cbar_kws`                    | Dictionary of keyword arguments for the color bar | `cbar_kws={'shrink': 0.8}`                    |
| `mask`                        | Mask certain cells (boolean 2D array)             | `mask=np.triu(corr)`                          |
| `ax`                          | Matplotlib axis object to draw the heatmap on     | `ax=ax`                                       |

---

#### Tips for Using Heatmaps:

* Use **correlation heatmaps** to detect multicollinearity or linear relationships.
* Use **pivot tables** for group-wise matrix comparisons (e.g. day vs. time, category vs. region).
* Use `mask` and `square=True` to make heatmaps cleaner and symmetrical for correlation matrices.
* Choose color palettes that reflect your data (e.g., diverging for correlation, sequential for quantity).



```python
corr = tips.corr(numeric_only=True)
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap (Tips Dataset)')
plt.show()
```

```python
pvflights = flights.pivot_table(values='passengers',index='month',columns='year',observed=False)
sns.heatmap(pvflights)
plt.show()
```

### Clustermap

A **clustermap** is a heatmap combined with **hierarchical clustering**. It not only visualizes a matrix of data but also automatically **reorders rows and columns** to group similar values together. This makes it extremely useful for **pattern discovery** in high-dimensional data.

In Seaborn, `clustermap()` builds on `heatmap()` but adds automatic clustering and dendrograms (tree diagrams) to highlight similarity across both rows and columns.

---

#### Clustermap Parameter Reference Table

| Parameter          | Description                                                    | Example Usage                      |
| ------------------ | -------------------------------------------------------------- | ---------------------------------- |
| `data`             | 2D matrix-like data (e.g. DataFrame or NumPy array)            | `data=corr`                        |
| `annot`            | Show data values in each cell                                  | `annot=True`                       |
| `fmt`              | String format for annotations                                  | `fmt=".2f"`                        |
| `cmap`             | Colormap for heatmap                                           | `cmap='viridis'`                   |
| `center`           | Value at which to center the colormap                          | `center=0`                         |
| `vmin` / `vmax`    | Min/max values for color scaling                               | `vmin=-1, vmax=1`                  |
| `row_cluster`      | Whether to perform hierarchical clustering on rows             | `row_cluster=True`                 |
| `col_cluster`      | Whether to perform hierarchical clustering on columns          | `col_cluster=True`                 |
| `row_linkage`      | Precomputed linkage matrix for rows                            | `row_linkage=linkage_matrix`       |
| `col_linkage`      | Precomputed linkage matrix for columns                         | `col_linkage=linkage_matrix`       |
| `z_score`          | Standardize rows (`0`) or columns (`1`)                        | `z_score=0`                        |
| `standard_scale`   | Normalize rows (`0`) or columns (`1`) to mean 0 and variance 1 | `standard_scale=1`                 |
| `figsize`          | Size of the figure                                             | `figsize=(10, 8)`                  |
| `dendrogram_ratio` | Proportions of dendrograms vs. heatmap                         | `dendrogram_ratio=(0.2, 0.2)`      |
| `cbar_pos`         | Color bar position (`x0`, `y0`, `width`, `height`)             | `cbar_pos=(0.02, 0.8, 0.03, 0.18)` |
| `colors_ratio`     | Space allocated for row/col colors                             | `colors_ratio=0.03`                |

---

### Tips for Using Clustermaps:

* Best for **exploring relationships** in large feature sets (e.g., correlation matrices, gene expression data).
* Use `z_score` or `standard_scale` to normalize data before clustering for better results.
* Use `row_cluster=False` or `col_cluster=False` to disable clustering on one axis.
* Works especially well when your data is **continuous** and high-dimensional.

---

```python
sns.clustermap(pvflights)
plt.show()
```

```python
sns.clustermap(pvflights,cmap='coolwarm',standard_scale=1)
plt.show()
```

## Boxenplot (Letter-value Plot)

A **boxenplot** (also known as a **letter-value plot**) is an enhancement of the traditional boxplot, designed to provide **more detail about the distribution's tails**. It is especially useful for **large datasets** where a boxplot might hide important variation beyond the interquartile range (IQR).

Unlike a boxplot, which typically shows just the median and quartiles, a boxenplot displays **progressively smaller subsets of the data**, giving more insight into the tails of the distribution.

Seaborn's `boxenplot()` is ideal when you want to understand **data spread, outliers, and tail behavior** in detail.

---

### Boxenplot Parameter Reference Table

| Parameter      | Description                                                           | Example Usage                         |
| -------------- | --------------------------------------------------------------------- | ------------------------------------- |
| `x` / `y`      | Categorical and numerical variables                                   | `x='day', y='total_bill'`             |
| `hue`          | Categorical variable for subgrouping                                  | `hue='sex'`                           |
| `data`         | DataFrame containing the input data                                   | `data=tips`                           |
| `order`        | Order of categorical axis levels                                      | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`    | Order for levels of hue variable                                      | `hue_order=['Female', 'Male']`        |
| `palette`      | Color palette to use                                                  | `palette='coolwarm'`                  |
| `orient`       | Orientation: `'v'` (vertical) or `'h'` (horizontal)                   | `orient='h'`                          |
| `width`        | Width of the boxes                                                    | `width=0.7`                           |
| `linewidth`    | Width of the box borders                                              | `linewidth=1.5`                       |
| `saturation`   | Saturation of box colors (0 to 1)                                     | `saturation=0.75`                     |
| `dodge`        | Separate boxes for hue levels                                         | `dodge=True`                          |
| `k_depth`      | Number of boxes to draw (e.g. `'profound'`, `'full'`, `'tukey'`, int) | `k_depth='profound'`                  |
| `scale`        | Method for box width: `'exponential'`, `'linear'`, etc.               | `scale='linear'`                      |
| `outlier_prop` | Proportion of data considered as outliers                             | `outlier_prop=0.007`                  |
| `ax`           | Matplotlib axis to draw the plot                                      | `ax=ax`                               |

---

### Tips for Using Boxenplots:

* Prefer `boxenplot` over `boxplot` when working with **large datasets**.
* Use `k_depth` to control how many letter-value boxes are drawn — `'profound'` for detailed tails, or fewer for simpler visuals.
* Combine with `hue` and `palette` to compare multiple subgroups visually.
* Try using it with **log-scale data** or **skewed distributions** to better explore asymmetry and outliers.


```python
sns.boxenplot(x='day', y='total_bill', data=tips)
plt.title('Boxenplot of Total Bill by Day')
plt.show()
```

```python
sns.boxenplot(x='day', y='total_bill', hue='sex', data=tips, palette='Set2')
plt.title('Boxenplot by Day and Sex')
plt.show()
```

## Pointplot

A **pointplot** is used to show **mean values** (or another estimator) of a numerical variable for different levels of one or more categorical variables. It also includes **confidence intervals**, making it great for visualizing **comparative trends** or **changes across categories**.

Unlike barplots, pointplots focus on **changes or comparisons**, not absolute values. They use **points connected by lines** to emphasize trends and relationships.

---

### Key Features:
- Plots **point estimates** (like mean) with **error bars** (like confidence intervals).
- Ideal for showing **trends across categories** or **within subgroups**.
- Can include a `hue` for subgroup comparisons.
- Supports customization of estimators, error bars, and point/line aesthetics.

---

### Pointplot Parameter Reference Table

| Parameter    | Description                                                | Example Usage                         |
| ------------ | ---------------------------------------------------------- | ------------------------------------- |
| `x` / `y`    | Categorical and numerical variables                        | `x='day', y='total_bill'`             |
| `hue`        | Categorical variable for subgrouping                       | `hue='sex'`                           |
| `data`       | DataFrame containing the input data                        | `data=tips`                           |
| `order`      | Order of the categories on the axis                        | `order=['Thur', 'Fri', 'Sat', 'Sun']` |
| `hue_order`  | Order of hue category levels                               | `hue_order=['Female', 'Male']`        |
| `palette`    | Color palette                                              | `palette='deep'`                      |
| `dodge`      | Offset the points for each hue level                       | `dodge=True`                          |
| `markers`    | Marker style for the points                                | `markers=['o', 's']`                  |
| `linestyles` | Line style connecting the points (`'-'`, `'--'`, etc.)     | `linestyles=['-', '--']`              |
| `capsize`    | Width of the caps on error bars                            | `capsize=0.1`                         |
| `errwidth`   | Thickness of the error bars                                | `errwidth=1.5`                        |
| `ci`         | Size of the confidence interval (`'sd'`, `None`, or int %) | `ci=95`                               |
| `estimator`  | Function to aggregate data (`mean`, `median`, etc.)        | `estimator=np.median`                 |
| `scale`      | Scaling factor for point size                              | `scale=1`                             |
| `orient`     | Orientation (`'v'` or `'h'`)                               | `orient='h'`                          |
| `join`       | Whether to connect points with a line                      | `join=True`                           |
| `ax`         | Matplotlib axis to draw the plot                           | `ax=ax`                               |

---

### Tips for Using Pointplots:

* Use `pointplot` to **compare averages or trends** across categories.
* Use `dodge=True` when plotting multiple subgroups with `hue` to prevent overlapping.
* Use `ci='sd'` or `ci=None` to customize error bars depending on your analysis needs.
* Best suited for **line-like comparisons**, not for distributions.


```python
sns.pointplot(x='day', y='total_bill', data=tips)
plt.title('Average Total Bill by Day')
plt.show()
```

```python
sns.pointplot(x='day', y='total_bill', hue='sex', data=tips, palette='muted', dodge=True)
plt.title('Average Total Bill by Day and Sex')
plt.show()
```
