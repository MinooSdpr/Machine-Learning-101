## Exploring Plot Types in Pandas

Pandas offers a rich variety of built-in plot types that are especially useful for quick statistical visualizations. These plots help you understand your data’s distribution, relationships, and trends with minimal effort.

Here are some of the most commonly used Pandas plot methods:

| Plot Type           | Description                                      | Method Example          |
|---------------------|------------------------------------------------|------------------------|
| Area Plot           | Displays cumulative totals over time or categories | `df.plot.area()`        |
| Horizontal Bar Plot | Bar plot with horizontal bars                    | `df.plot.barh()`        |
| Density Plot        | Estimates the probability density function       | `df.plot.density()` or `df.plot.kde()` |
| Histogram           | Shows the distribution of a numeric variable     | `df.plot.hist()`        |
| Line Plot           | Plots data points connected by lines             | `df.plot.line()`        |
| Scatter Plot        | Displays values as points for two variables      | `df.plot.scatter(x, y)` |
| Vertical Bar Plot   | Classic bar chart                                 | `df.plot.bar()`         |
| Box Plot            | Visualizes data spread and outliers               | `df.plot.box()`         |
| Hexbin Plot         | Shows bivariate data density with hexagons       | `df.plot.hexbin(x, y, gridsize=30)` |
| Pie Chart           | Displays data as slices of a pie                   | `df.plot.pie()`         |


### Area Plot

An **area plot** is essentially a line plot with the area below the line filled in. It is useful for visualizing cumulative data and comparing multiple data series over time or categories. Area plots can help emphasize the magnitude of change between variables.


#### Key Parameters:

* **alpha** (`float`): Controls the transparency of the filled area. A value between 0 and 1 (default is `1.0`). Setting it around `0.5` helps when plotting multiple series.
* **stacked** (`bool`): If `True` (default), plots stacked areas. If `False`, plots each series independently.
* **figsize** (`tuple`): Specifies the size of the figure in inches.
* **title**, **xlabel**, **ylabel**: Standard matplotlib-style labels can also be added for better readability.

```python
df2.plot.area(alpha=0.4)
plt.show()
```

### Barplots

A **bar plot** is one of the most common types of categorical data visualizations. It displays data with rectangular bars representing the magnitude of each category. In Pandas, `df.plot.bar()` creates vertical bar plots, while `df.plot.barh()` creates horizontal ones.


#### Key Parameters:

| Parameter    | Description                                                                                   |
| ------------ | --------------------------------------------------------------------------------------------- |
| **x**        | Column name to use as labels (optional if DataFrame is already indexed appropriately)         |
| **y**        | Column(s) to plot (can be a string or list of strings)                                        |
| **stacked**  | `True` to stack multiple columns on top of each other; `False` (default) to plot side-by-side |
| **color**    | Color(s) of the bars. Accepts a single color, a list of colors, or a color map                |
| **width**    | Width of the bars (default is `0.8`)                                                          |
| **figsize**  | Tuple for figure size in inches, e.g., `(10, 6)`                                              |
| **title**    | Title of the plot                                                                             |
| **legend**   | Shows or hides legend (`True` by default)                                                     |
| **rot**      | Rotation angle for x-axis labels                                                              |
| **fontsize** | Font size for tick labels                                                                     |
| **grid**     | Shows grid if set to `True`                                                                   |
| **position** | Position of bars when plotting multiple columns                                               |
| **align**    | Alignment of bars relative to the tick labels (`'center'` or `'edge'`)                        |



```python
df2.head()
```

```python
df2.plot.bar()
plt.show()
```

```python
df2.plot.bar(stacked=True)
plt.show()
```

```python
df2[['a', 'b']].plot.bar(
    stacked=True,
    color=['skyblue', 'orange'],
    width=0.6,
    figsize=(10, 6),
    title='Stacked Bar Plot Example',
    rot=45,
    grid=True
)
plt.show()
```

### Histograms

A **histogram** is a powerful plot for understanding the distribution of numerical data. It groups values into bins and displays the frequency (or count) of values falling into each bin as bars. In Pandas, `df.plot.hist()` or `df['column'].hist()` is used to create histograms.

#### Key Parameters:

| Parameter                       | Description                                                   |
| ------------------------------- | ------------------------------------------------------------- |
| **bins**                        | Number of bins or a list of bin edges (default is `10`)       |
| **figsize**                     | Tuple specifying the figure size in inches                    |
| **density**                     | If `True`, draws probability densities instead of frequencies |
| **alpha**                       | Transparency level of the bars (0.0 to 1.0)                   |
| **color**                       | Color of the bars                                             |
| **grid**                        | Show/hide grid lines (`True` by default)                      |
| **xlabelsize** / **ylabelsize** | Font size of axis labels                                      |
| **xrot** / **yrot**             | Rotation of tick labels on x/y axes                           |
| **cumulative**                  | If `True`, shows a cumulative histogram                       |
| **legend**                      | Whether to show the legend                                    |

```python
df1['A'].plot.hist(bins=50)
plt.show()
```

```python
df1['A'].hist(
    bins=20,
    figsize=(8, 5),
    color='teal',
    alpha=0.7,
    grid=True,
    xlabelsize=12,
    ylabelsize=12
)
plt.show()
```

### Line Plots

The `line()` method in Pandas is a convenient wrapper around `df.plot(kind='line')` and is used to create **line plots**, which are ideal for visualizing trends, patterns, and changes over continuous intervals—especially time series data.


#### Key Parameters:

| Parameter      | Description                                                         |
| -------------- | ------------------------------------------------------------------- |
| **x**          | Column label for x-axis (optional). If not specified, index is used |
| **y**          | Column label or list of labels to plot on the y-axis                |
| **figsize**    | Size of the figure as a tuple `(width, height)`                     |
| **linewidth**  | Thickness of the line (default is `1.5`)                            |
| **linestyle**  | Style of the line (`'-'`, `'--'`, `':'`, `'-.'`, etc.)              |
| **marker**     | Marker symbol at data points (e.g., `'o'`, `'x'`, `'^'`)            |
| **color**      | Line color(s) as a string or list                                   |
| **grid**       | Display grid lines (`True` or `False`)                              |
| **title**      | Title of the plot                                                   |
| **alpha**      | Transparency of the lines (from `0.0` to `1.0`)                     |
| **legend**     | Show or hide the legend (`True` by default)                         |
| **rot**        | Rotation of x-axis tick labels                                      |
| **fontsize**   | Font size of tick labels                                            |
| **use\_index** | Whether to use DataFrame index for the x-axis (`True` by default)   |

```python
df1.reset_index(inplace=True)

df1.head()
```

```python
df1.plot.line(x='index', y = 'B',figsize=(12,3),lw=1)
plt.show()
```

```python
df1.plot.line(
    y=['A', 'B'],
    figsize=(10, 5),
    linewidth=2,
    linestyle='--',
    marker='o',
    color=['blue', 'orange'],
    title='Line Plot of A and B',
    grid=True
)
plt.show()
```

### Scatter Plots

The `scatter()` method in Pandas creates a **scatter plot**, which is perfect for visualizing the relationship or correlation between two numeric variables. Each point represents an observation, plotted according to its values on the x and y axes.

#### Key Parameters:

| Parameter    | Description                                                                                 |
| ------------ | ------------------------------------------------------------------------------------------- |
| **x**        | Column name for the x-axis values (required)                                                |
| **y**        | Column name for the y-axis values (required)                                                |
| **s**        | Size of the markers. Can be a scalar or an array-like (e.g., `df['size']`)                  |
| **c**        | Color of the markers. Can be a single color or an array-like of values mapped to a colormap |
| **figsize**  | Tuple specifying figure size in inches, e.g., `(8, 6)`                                      |
| **colorbar** | If `True`, shows a colorbar when `c` is array-like (default is `False`)                     |
| **title**    | Title of the plot                                                                           |
| **grid**     | Show or hide grid lines (default is `True`)                                                 |
| **alpha**    | Transparency level of markers (0.0 transparent through 1.0 opaque)                          |
| **marker**   | Marker style (e.g., `'o'`, `'x'`, `'^'`)                                                    |
| **kwargs**   | Additional keyword arguments passed to Matplotlib’s `scatter()`                             |


```python
df1.plot.scatter(x='A',y='B')
plt.show()
```

You can use c to color based off another column value
Use cmap to indicate colormap to use. 
For all the colormaps, check out: https://matplotlib.org/stable/users/explain/colors/colormaps.html

```python
df1.plot.scatter(x='A',y='B',c='C',cmap='coolwarm')
plt.show()
```

Or use s to indicate size based off another column. s parameter needs to be an array, not just the name of a column:

```python
df1['C'] = df1['C'].apply(lambda x: 0 if x<0 else x)
```

```python
df1.plot.scatter(
    x='A',
    y='B',
    s=df1['C'].fillna(1) * 20,     
    c='D',                
    colormap='viridis',
    figsize=(9, 6),
    alpha=0.7,
    grid=True)
plt.show()
```

### Box Plots

The `box()` method in Pandas creates a **box plot** (also known as a whisker plot), which is useful for visualizing the distribution of data through their quartiles. Box plots highlight the median, spread, and potential outliers of numerical data, making them excellent for comparing distributions across multiple columns or groups.

#### Key Parameters:

| Parameter                                                                                    | Description                                                                               |
| -------------------------------------------------------------------------------------------- | ----------------------------------------------------------------------------------------- |
| **by**                                                                                       | Column name to group data by before plotting multiple box plots                           |
| **figsize**                                                                                  | Tuple specifying figure size in inches, e.g., `(8, 6)`                                    |
| **grid**                                                                                     | Show or hide grid lines (`True` by default)                                               |
| **color**                                                                                    | Dict to specify colors for different elements like boxes, whiskers, medians, etc.         |
| **vert**                                                                                     | If `True` (default), plot vertical boxes; if `False`, horizontal boxes                    |
| **patch\_artist**                                                                            | If `True`, fills the box with color                                                       |
| **showmeans**                                                                                | If `True`, displays the mean of each distribution                                         |
| **meanline**                                                                                 | If `True` and `showmeans=True`, draws the mean as a line instead of a point               |
| **showcaps**                                                                                 | Show or hide the end lines of whiskers                                                    |
| **showbox**                                                                                  | Show or hide the main box                                                                 |
| **showfliers**                                                                               | Show or hide outliers (points outside whiskers)                                           |
| **notch**                                                                                    | If `True`, draws a notched box to compare medians                                         |
| **whiskerprops**, **capprops**, **boxprops**, **flierprops**, **meanprops**, **medianprops** | Dictionaries to customize the appearance of respective plot elements                      |


```python
df2.plot.box() # Can also pass a by= argument for groupby
plt.show()
```

```python
df2.plot.box(
    figsize=(10, 6),
    grid=True,
    vert=True,
    patch_artist=True,
    showmeans=True,
    meanline=True,
    color=dict(boxes='LightBlue', whiskers='DarkBlue', medians='Red', caps='Black'),
    title='Box Plot of Columns A, B, and C'
)
plt.show()
```

### Hexagonal Bin Plot

A **hexagonal bin plot** is a great way to visualize the density of points when you have large datasets with overlapping scatter points. Instead of plotting every point, the plot aggregates points into hexagonal bins, with color intensity representing the number of points in each bin.

#### Key Parameters:

| Parameter               | Description                                                          |
| ----------------------- | -------------------------------------------------------------------- |
| **x**                   | Column name for x-axis values (required)                             |
| **y**                   | Column name for y-axis values (required)                             |
| **gridsize**            | Number of hexagons in the x-direction (default is 30)                |
| **figsize**             | Tuple specifying figure size `(width, height)` in inches             |
| **cmap**                | Colormap to use for hexagons (default is `'viridis'`)                |
| **reduce\_C\_function** | Function to aggregate values in each hexagon (default is `np.sum`)   |
| **mincnt**              | Minimum count of points in a hexagon to be colored (default is None) |
| **edgecolors**          | Color of hexagon edges (default is `'none'` to avoid lines)          |
| **linewidths**          | Width of hexagon edges (default is 0)                                |
| **title**               | Plot title                                                           |

```python
df = pd.DataFrame(np.random.randn(1000, 2), columns=['a', 'b'])
df.plot.hexbin(x='a',y='b',gridsize=25,cmap='Oranges')
plt.show()
```

```python
df1.plot.hexbin(
    x='A',
    y='B',
    gridsize=40,
    cmap='plasma',
    figsize=(10, 6),
    reduce_C_function=np.mean,
    edgecolors='grey',
    linewidths=0.5,
    title='Hexagonal Bin Plot of A vs B'
)
plt.show()
```

### Kernel Density Estimation Plot (KDE)

A **Kernel Density Estimation (KDE) plot** is a smooth, continuous probability density curve that estimates the distribution of a numerical variable. It’s a great alternative to histograms when you want a smooth visualization of the data distribution.

#### Key Parameters:

| Parameter              | Description                                                                                                                                                     |
| ---------------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **bw\_method**         | Bandwidth method for smoothing. Can be a scalar or string (e.g., `'scott'`, `'silverman'`). Smaller values produce a more detailed curve. Default is automatic. |
| **ind**                | Points at which to evaluate the KDE (array-like). Default generates 1000 points.                                                                                |
| **figsize**            | Tuple specifying figure size in inches, e.g., `(8, 6)`                                                                                                          |
| **gridsize**           | Number of points to evaluate KDE over the data range (default 1000)                                                                                             |
| **legend**             | Show or hide legend (`True` by default)                                                                                                                         |
| **xlabel**, **ylabel** | Labels for x and y axes                                                                                                                                         |
| **xlim**, **ylim**     | Limits for x and y axes                                                                                                                                         |
| **color**              | Color(s) of the KDE lines                                                                                                                                       |
| **linestyle**          | Line style (e.g., `'-'`, `'--'`)                                                                                                                                |
| **linewidth**          | Thickness of the KDE line (default 1.5)                                                                                                                         |
| **alpha**              | Transparency of the KDE lines (0.0 to 1.0)                                                                                                                      |
| **log\_scale**         | If `True`, applies log scale to the axes                                                                                                                        |
| **kwargs**             | Additional keyword arguments passed to matplotlib plotting functions                                                                                            |

```python
df2['a'].plot.kde()
plt.show()
```

```python
df2.plot.density()
plt.show()
```

```python
df1[['A', 'B']].plot.kde(
    bw_method=0.3,
    figsize=(10, 6),
    linewidth=2,
    linestyle='-',
    alpha=0.8,
    grid=True
)
plt.show()
```
