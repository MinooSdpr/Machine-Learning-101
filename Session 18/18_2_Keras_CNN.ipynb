```python
y_cat_test = to_categorical(y_test,10)
```

### Processing X Data

We should normalize the X data

```python
x_train = x_train/255
x_test = x_test/255
```

## Reshaping the Data

We need to add one more dimension to show we're dealing with 1 RGB channel (since technically the images are in black and white, only showing values from 0-255 on a single channel), an color image would have 3 dimensions.

```python
x_test.shape
```

Reshape to include channel dimension (in this case, 1 channel)

```python
x_train = x_train.reshape(60000, 28, 28, 1)
```

```python
x_train.shape
```

```python
x_test = x_test.reshape(10000,28,28,1)
```

```python
x_test.shape
```

## CNN

In a Convolutional Neural Network (CNN), the core building blocks are:

* **`Conv2D`**: Applies convolution filters to extract features like edges, textures, etc.
* **`MaxPooling2D`**: Reduces the spatial size of the feature maps, keeping the most important information.
* **`Flatten`**: Converts multi-dimensional tensors into 1D vectors to feed into fully connected layers.

---

### 🧱 Layer Descriptions

| **Layer**      | **Description**                                                                                                               |
| -------------- | ----------------------------------------------------------------------------------------------------------------------------- |
| `Conv2D`       | Applies a set of convolutional filters (kernels) to the input. Each filter extracts a different feature from the input image. |
| `MaxPooling2D` | Downsamples the input by taking the maximum value over a window (usually 2x2), reducing dimensionality and computation.       |
| `Flatten`      | Flattens the input tensor into a 1D vector, used to transition from convolutional layers to dense (fully connected) layers.   |

---

### ⚙️ Common Parameters

| **Layer**      | **Parameter** | **Description**                                                            |
| -------------- | ------------- | -------------------------------------------------------------------------- |
| `Conv2D`       | `filters`     | Number of output filters (feature maps)                                    |
|                | `kernel_size` | Size of the convolution window (e.g., `(3, 3)`)                            |
|                | `strides`     | Step size for moving the kernel                                            |
|                | `padding`     | `'valid'` (no padding) or `'same'` (zero-padding to keep same output size) |
|                | `activation`  | Activation function, e.g., `'relu'`, `'sigmoid'`, etc.                     |
|                | `input_shape` | Shape of input (used only in the first layer), e.g., `(28, 28, 1)`         |
| `MaxPooling2D` | `pool_size`   | Size of the pooling window (e.g., `(2, 2)`)                                |
|                | `strides`     | Step size for the pool window                                              |
|                | `padding`     | `'valid'` or `'same'`                                                      |
| `Flatten`      | —             | No parameters; it just reshapes the tensor                                 |



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, MaxPool2D, Flatten, Dropout
```

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPool2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPool2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model.summary()
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)
```

## Train the Model

### 🧠 What is `validation_split`?

* `validation_split=0.1` means:

  * **10% of the training data** will be used as a **validation set**
  * **90% will be used to train the model**

### 🧪 Why use a validation split?

* The **validation set** is used to evaluate the model's performance **after each epoch**, but **not used to update the weights**.
* This allows the model to be monitored for **overfitting** or **underfitting**.
* With `EarlyStopping`, the training will stop if the validation loss doesn't improve for a few epochs (`patience=2` in your case).


```python
history = model.fit(
    x_train, y_cat_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    callbacks=[early_stop]
)
```

## Evaluate the Model

```python
model.metrics_names
```

```python
losses = pd.DataFrame(model.history.history)
```

```python
losses.head()
```

```python
losses[['accuracy','val_accuracy']].plot()
```

```python
losses[['loss','val_loss']].plot()
```

```python
print(model.metrics_names)
print(model.evaluate(x_test,y_cat_test,verbose=0))
```
