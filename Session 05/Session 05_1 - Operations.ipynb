## 🔍 Exploring Unique Values

Understanding the **distinct values** in a column is crucial for analysis.

### Get unique values in a column:

```python
import pandas as pd

df = pd.DataFrame({
    'col1': [1, 2, 3, 4],
    'col2': [444, 555, 666, 444],
    'col3': ['abc', 'def', 'ghi', 'xyz']
})

df.head()
```

```python
df['col2'].unique()
```

### Count of unique values:

```python
df['col2'].nunique()
```

### Frequency of each value (value counts):

```python
df['col2'].value_counts()
```

## 🗑️ Permanently Remove a Column

To **delete a column** from the DataFrame:

```python
del df['col1']
```

```python
df
```

## 📊 Sorting & Ordering the DataFrame

You can sort a DataFrame by column values:

> ℹ️ By default, `inplace=False`, meaning the original DataFrame remains unchanged unless specified otherwise.

```python
df.sort_values(by='col2') #inplace=False by default
```

## 🕵️‍♂️ Detecting Missing or Null Values

To **check for null (missing) values**:

> ✅ This is useful for identifying data that may require cleaning.

```python
df.isnull()
```
