## Using Bar Charts
| Argument     | Description                                                                                      | Default/Type                            |               |
| ------------ | ------------------------------------------------------------------------------------------------ | --------------------------------------- | ------------- |
| `x`          | The x coordinates of the bars (categorical or numeric).                                          | Required                                |               |
| `height`     | The height(s) of the bars (values for each bar).                                                 | Required                                |               |
| `width`      | The width(s) of the bars.                                                                        | Default: 0.8                            |               |
| `bottom`     | The y coordinate(s) of the bottom of the bars (for stacked bars).                                | Default: 0                              |               |
| `align`      | Alignment of the bars to the x coordinates: `'center'` or `'edge'`.                              | Default: `'center'`                     |               |
| `color`      | Color or sequence of colors for the bars.                                                        | Default: None (uses cycle)              |               |
| `edgecolor`  | Color of the bar edges.                                                                          | Default: None                           |               |
| `linewidth`  | Width of the bar edges.                                                                          | Default: None                           |               |
| `tick_label` | Labels for the bars on the x-axis.                                                               | Default: None                           |               |
| `xerr`       | Error values for bars in the x-direction (scalar or array).                                      | Default: None                           |               |
| `yerr`       | Error values for bars in the y-direction (scalar or array).                                      | Default: None                           |               |
| `ecolor`     | Color of the error bars.                                                                         | Default: `'black'`                      |               |
| `capsize`    | Length of the error bar caps in points.                                                          | Default: 0                              |               |
| `error_kw`   | Dictionary of kwargs for the error bars (e.g., `elinewidth`, `alpha`).                           | Default: None                           |               |
| `log`        | If True, sets the y-axis to a logarithmic scale.                                                 | Default: False                          |               |
| `label`      | Label for the bars (used in legends).                                                            | Default: None                           |               |
| `hatch`      | Pattern for filling bars (e.g., '/', '\\', '                                                     | ', '-', '+', 'x', 'o', 'O', '.', '\*'). | Default: None |
| `data`       | An optional parameter to pass a data structure (like DataFrame) for referencing columns by name. | Default: None                           |               |



```python
x = ['Nuclear', 'Hydro', 'Coal', 'Gas', 'Solar', 'Wind', 'Other']
per_1 = [71, 10, 3, 7, 2, 4, 3]
variance = [8, 3, 1, 3, 1, 2, 1]
# barh makes horizontal chart
plt.bar(x, per_1, color='purple', yerr=variance, ecolor = 'red')
plt.show()
```

```python
m_eng = (76, 85, 86, 88, 93)
f_eng = (24, 15, 14, 12, 7)

spc = np.arange(5)

plt.bar(spc, m_eng, width=0.45, label='Male', edgecolor='k')
plt.show()
```

```python
plt.xticks(spc + 0.45 / 2, ('Aero', 'Chem', 'Civil', 'Elec', 'Mech'))
plt.bar(spc + 0.45, f_eng, width=0.45, label='Female', edgecolor='k')
plt.show()
```

```python
t_type = ['Kind', 'Elem', 'Sec', 'Spec']
m_teach = np.array([2, 20, 44, 14])
f_teach = np.array([98, 80, 56, 86])
ind = [x for x, _ in enumerate(t_type)]

plt.bar(ind, m_teach, width=0.45, label='Male', bottom=f_teach)
plt.bar(ind, f_teach, width=0.45, label='Female')

plt.legend(loc='center')
plt.show()
```

## Pie Chart with Custom Styling

| Argument        | Description                                                                          |
| --------------- | ------------------------------------------------------------------------------------ |
| `x`             | **(Required)** Sequence of numerical values (e.g., \[20, 30, 50]) to plot as slices. |
| `labels`        | Labels for each slice. Example: `['A', 'B', 'C']`.                                   |
| `colors`        | List of colors for the slices. Can be names, hex codes, or RGB tuples.               |
| `explode`       | A sequence of values (e.g., `[0.1, 0, 0]`) that offsets slices outward for emphasis. |
| `autopct`       | A string or function to label slices with their numeric value. Example: `'%1.1f%%'`. |
| `shadow`        | Adds a shadow behind the pie for depth. Boolean.                                     |
| `startangle`    | Rotates the start of the pie chart in degrees. Useful for reorienting.               |
| `radius`        | Radius of the pie. Default is `1`.                                                   |
| `counterclock`  | If `True`, slices are ordered counterclockwise. Default: `True`.                     |
| `pctdistance`   | Distance of the `autopct` text from the center (float).                              |
| `labeldistance` | Distance of the labels from the center (float).                                      |
| `frame`         | If `True`, draws a frame (circle) around the pie chart.                              |
| `normalize`     | If `True`, forces data to sum to 1. Default is `None`.                               |
| `wedgeprops`    | Dictionary of properties for the wedges (e.g., line width, edge color).              |
| `textprops`     | Dictionary of properties for the text labels (e.g., font size, color).               |
| `center`        | Tuple of (x, y) for pie center position. Default is `(0, 0)`.                        |
| `rotatelabels`  | If `True`, rotates each label to match the slice angle.                              |
| `data`          | When using DataFrames, pass a reference to columns via `data=...`.                   |


```python
import random

fig_6 = plt.figure(figsize=(8,5),dpi=100)
axes_6 = fig_6.add_axes([0.1,0.1,0.9,0.9])

types = ['Water', 'Normal', 'Flying', 'Grass', 'Psychic', 'Bug', 'Fire', 'Poison',
'Ground', 'Rock', 'Fighting', 'Dark', 'Steel', 'Electric', 'Dragon', 'Fairy',
'Ghost', 'Ice']
poke_num = [133, 109, 101, 98, 85, 77, 68, 66, 65, 60, 57, 54, 53, 51, 50, 50, 46, 40]

colors = []
for i in range(18):
    rgb = (random.uniform(0, .5), random.uniform(0, .5), random.uniform(0, .5))
    colors.append(rgb)
    
# Explode biggest 1st slice
explode = [0] * 18
explode[0] = 0.15


wedges, texts, autotexts = plt.pie(poke_num, explode=explode, labels=types, colors=colors, autopct='%1.0f%%', 
                                   shadow=True, startangle=120,  textprops=dict(color="w"))
# Create legend to right and move off pie with 1-1.5 axes point width
plt.legend(wedges, types, loc='right', bbox_to_anchor=(1, 0, 0.5, 1)) #(x, y, width, height) or (x, y)
plt.show()
```

## 📈 Visualizing with line chart Over Time

### 📅 Handling Business Days with `pd.bdate_range`

- **`pd.bdate_range()`** is used to generate a **range of business days** (excluding weekends and custom holidays) between two dates.
- Parameters:
  - `start`: Starting date 
  - `end`: Ending date
  - `holidays`: A list of custom holiday dates that should be excluded (e.g., May 25 and Aug 19, 2020).

#### 🕒 Common `freq` Values in `pandas`

##### 📆 **Calendar Frequencies**

| Code                       | Description                | Example                     |
| -------------------------- | -------------------------- | --------------------------- |
| `'D'`                      | Calendar day               | 2025-01-01, 2025-01-02, ... |
| `'W'`                      | Weekly (default: Sunday)   | 2025-01-05, 2025-01-12, ... |
| `'W-MON'`, `'W-FRI'`, etc. | Weekly on specific weekday | 2025-01-06 (Monday), etc.   |
| `'M'`                      | Month end                  | 2025-01-31, 2025-02-29, ... |
| `'MS'`                     | Month start                | 2025-01-01, 2025-02-01, ... |
| `'Q'`                      | Quarter end                | 2025-03-31, 2025-06-30, ... |
| `'QS'`                     | Quarter start              | 2025-01-01, 2025-04-01, ... |
| `'A'` or `'Y'`             | Year end                   | 2025-12-31, 2026-12-31, ... |
| `'AS'` or `'YS'`           | Year start                 | 2025-01-01, 2026-01-01, ... |

---

##### 💼 **Business Frequencies**

| Code               | Description                                |
| ------------------ | ------------------------------------------ |
| `'B'`              | Business day (Mon–Fri, excluding weekends) |
| `'C'`              | Custom business day (with holidays)        |
| `'BM'`             | Business month end                         |
| `'BMS'`            | Business month start                       |
| `'BQ'`             | Business quarter end                       |
| `'BQS'`            | Business quarter start                     |
| `'BA'` or `'BY'`   | Business year end                          |
| `'BAS'` or `'BYS'` | Business year start                        |

---

##### 🕰️ **Time-Based Frequencies**

| Code             | Description |
| ---------------- | ----------- |
| `'H'`            | Hour        |
| `'T'` or `'min'` | Minute      |
| `'S'`            | Second      |
| `'L'` or `'ms'`  | Millisecond |
| `'U'` or `'us'`  | Microsecond |
| `'N'`            | Nanosecond  |


```python
import datetime

goog_data = pd.read_csv('GOOG.csv')
goog_data_np = goog_data.to_numpy()

goog_cp = goog_data_np[:,4]
goog_cp

holidays = [datetime.datetime(2020,5,25), datetime.datetime(2020,8,19)]
date_arr = pd.bdate_range(start='5/20/2020', end='8/19/2020', freq='C', holidays=holidays)
date_arr_np = date_arr.to_numpy()

fig_7 = plt.figure(figsize=(8,5),dpi=100)
axes_7 = fig_7.add_axes([0.1,0.1,0.9,0.9])
plt.plot(date_arr_np, goog_cp)
plt.show()
```

## 🔹 **Data & Content**

| Argument      | Description                                                                         |
| ------------- | ----------------------------------------------------------------------------------- |
| `cellText`    | 2D list or array of strings or numbers to fill the table cells (excluding headers). |
| `cellColours` | 2D list of colors for each cell. Must match the shape of `cellText`.                |

---

### 🔸 **Headers**

| Argument     | Description                                   |
| ------------ | --------------------------------------------- |
| `rowLabels`  | List of row labels (shown on the left side).  |
| `colLabels`  | List of column headers (shown on top).        |
| `rowColours` | List of background colors for row labels.     |
| `colColours` | List of background colors for column headers. |

---

### 🧭 **Positioning & Alignment**

| Argument  | Description                                                                                                                     |
| --------- | ------------------------------------------------------------------------------------------------------------------------------- |
| `cellLoc` | Alignment of text in each cell: `'left'`, `'center'`, or `'right'`. Default is `'right'`.                                       |
| `rowLoc`  | Vertical alignment of row labels.                                                                                               |
| `colLoc`  | Horizontal alignment of column headers.                                                                                         |
| `loc`     | Location of the table in the axes: `'bottom'`, `'top'`, `'left'`, `'right'`, `'center'`, etc.                                   |
| `bbox`    | A tuple/list `(x0, y0, width, height)` to manually place the table (coordinates in axes fraction, not pixels). Overrides `loc`. |

---

### 🧱 **Layout**

| Argument    | Description                                                                                                           |
| ----------- | --------------------------------------------------------------------------------------------------------------------- |
| `colWidths` | List of column widths (fractions of the axes width). Example: `[0.2, 0.4, 0.4]`.                                      |
| `edges`     | Which edges of the table to draw. Default is `'closed'`. Options: `'open'`, `'horizontal'`, `'vertical'`, `'closed'`. |

```python
goog_data['Open'] = pd.Series([round(val, 2) for val in goog_data['Open']], 
                              index = goog_data.index)
goog_data['High'] = pd.Series([round(val, 2) for val in goog_data['High']], 
                              index = goog_data.index)
goog_data['Low'] = pd.Series([round(val, 2) for val in goog_data['Low']], 
                              index = goog_data.index)
goog_data['Close'] = pd.Series([round(val, 2) for val in goog_data['Close']], 
                              index = goog_data.index)
goog_data['Adj Close'] = pd.Series([round(val, 2) for val in goog_data['Adj Close']], 
                              index = goog_data.index)

stk_data = goog_data[-5:]
stk_data
```

```python
col_head = ('Date','Open','High','Low','Close','Adj Close','Volume')

stk_data_np = stk_data.to_numpy()
stk_data_np
```

```python
plt.figure(linewidth=2,figsize=(5,3))

axes_8 = plt.gca() #get current axes
axes_8.get_xaxis().set_visible(False)
axes_8.get_yaxis().set_visible(False)
plt.box(on=None)

# np.full returns an array filled with 0.1
# cm is a colormap object we are using to use a default blue color
ccolors = plt.cm.Blues(np.full(len(col_head), 0.5))

the_table = plt.table(cellText=stk_data_np, loc='center', colLabels=col_head,
                     colColours=ccolors )
the_table.set_fontsize(14)
the_table.scale(3, 2.5)
plt.show()
```

## Scatter Plot

| Argument       | Description                                                                                                                    |
| -------------- | ------------------------------------------------------------------------------------------------------------------------------ |
| `x`            | Array-like – X coordinates of the data points                                                                                  |
| `y`            | Array-like – Y coordinates of the data points                                                                                  |
| `s`            | Scalar or array – Size of the markers (points). Default is 20.                                                                 |
| `c`            | Color or sequence of colors – Marker color(s). Can be a string (e.g., `'red'`), hex (`'#ff0000'`), or array for color mapping. |
| `marker`       | Marker style – e.g., `'o'` (circle), `'^'` (triangle), `'s'` (square), etc.                                                    |
| `cmap`         | Colormap – Used when `c` is numeric; applies a color gradient (e.g., `'viridis'`, `'plasma'`, `'Blues'`).                      |
| `norm`         | Normalize object – Custom normalization for color scale.                                                                       |
| `vmin`, `vmax` | Min/max values for color mapping when `c` is an array.                                                                         |
| `alpha`        | Float – Opacity (0 = transparent, 1 = solid).                                                                                  |
| `linewidths`   | Width of the marker edge lines.                                                                                                |
| `edgecolors`   | Color of marker edges. Default is `'face'` (same as fill color), or `'none'`, `'black'`, etc.                                  |
| `label`        | String – Label for the legend.                                                                                                 |


```python
cnt_arr = np.array(['Australia','Brazil','Canada','Chile','France','Germany','Greece',
                   'Iceland','India','Iran','Italy','Mexico','New Zealand','Nigeria',
                   'Norway','Pakistan','Peru','Russia','Saudi Arabia','Singapore',
                   'South Africa','Spain','Sweden','Turkey','UK','US'])
dr_arr = np.array([1.8,53,24.5,56.5,45.4,11.2,2.2,
                   2.8,4,24.6,58.6,46.3,.5,.5,
                   4.9,2.9,83.3,11,10.4,.5,
                   21.5,61.6,56.9,7.3,62.4,52.9])

test_arr = np.array([110,7197,600,1862,1636,1103,35,
                   10,295,1658,1226,2490,8,243,
                   48,1395,1101,4447,1443,280,
                   2830,1602,447,1205,1546,24988])

cc_arr = np.array([24236,3456652,125408,390037,256534,229706,7684,
                   2035,2836925,350279,255278,537031,1654,50488,
                   10162,290445,549321,935066,302686,56031,
                   596060,370867,85411,253108,323008,5529824])
cc_arr_sm = cc_arr / 1000
color_arr = np.random.rand(26)

plt.title('Death Rate per 100k vs. Confirmed Cases')
plt.xlabel('Death Rate per 100k')
plt.ylabel('Confirmed Cases')
plt.scatter(dr_arr,test_arr,s=cc_arr_sm,c=color_arr,alpha=0.5)
plt.show()
```

## 📍 3D Scatter Plot

| Argument       | Type                 | Description                                                           |
| -------------- | -------------------- | --------------------------------------------------------------------- |
| `x`      | array-like | X coordinates of points |
| `y`      | array-like | Y coordinates of points |
| `z`      | array-like | Z coordinates of points |
| `c`            | array-like or color  | Values to map to colors (colormapped if array, solid if color string) |
| `cmap`         | string or Colormap   | Name of colormap (`'viridis'`, `'Blues'`, etc.)                       |
| `norm`         | Normalize object     | Normalize the colormap scaling                                        |
| `vmin`, `vmax` | float                | Value limits for color mapping                                        |
| `alpha`        | float (0.0–1.0)      | Transparency level of the markers                                     |
| `s`            | scalar or array      | Marker size (in points²)                                              |
| `marker`       | string               | Marker style (e.g., `'o'`, `'^'`, `'x'`, etc.)                        |
| `edgecolors`   | color or sequence    | Edge color of the markers                                             |
| `linewidths`   | float or array       | Width of marker edge lines                                            |
| `label`        | string               | Label for the legend                                                  |
| `depthshade`   | bool (default: True) | If True, darker markers appear closer                                 |
| `picker`       | float or bool        | Enable picking for interactivity                                      |
| `zorder`       | int                  | Drawing order (higher values appear on top)                           |
| `rasterized`   | bool                 | Render markers as raster graphics                                     |



```python
from mpl_toolkits import mplot3d

fig_9 = plt.figure(figsize=(8,5),dpi=100)
axes_9 = fig_9.add_axes([0.1,0.1,0.9,0.9], projection='3d')

# Create a 3D scatterplot
# Te darker points are represented that way to seem closer to you
z_3 = 15 * np.random.random(100)
x_3 = np.sin(z_3) * np.random.randn(100)
y_3 = np.cos(z_3) * np.random.randn(100)
axes_9.scatter3D(x_3, y_3, z_3, c=z_3, cmap='Blues')
plt.show()
```

```python
fig_9 = plt.figure(figsize=(9,5),dpi=100)
axes_9 = fig_9.add_axes([0,0,1,1], projection='3d')

def get_z(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

x_4 = np.linspace(-6, 6, 30)
y_4 = np.linspace(-6, 6, 30)

x_4, y_4 = np.meshgrid(x_4, y_4)
z_4 = get_z(x_4,y_4)

# Change viewing angle to reorient camera 45 degrees and rotate 55
axes_9.view_init(45,55)

# Provide x, y, z, contours and color map
axes_9.contour3D(x_4,y_4,z_4,cmap='Blues')
axes_9.set_xlabel('x')
axes_9.set_ylabel('y')
axes_9.set_zlabel('z')

# You can create wireframes
axes_9.plot_wireframe(x_4,y_4,z_4,color='green')

axes_9.plot_surface(x_4,y_4,z_4, rstride=1, cstride=1,cmap='Blues',edgecolor='none')
plt.show()
```
